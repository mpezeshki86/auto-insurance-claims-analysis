# Business and Actuarial Insights

This section translates the exploratory analysis and claim-frequency modeling results into actuarial and business insights. The objective is to identify meaningful risk patterns, evaluate their potential relevance for insurance segmentation and pricing analysis, and summarize the practical implications and limitations of the findings.

## Executive Summary

This analysis identifies meaningful differences in auto insurance claim frequency across several risk characteristics. BonusMalus is the strongest modeled risk indicator, with the highest category exhibiting approximately 5.17 times the modeled claim frequency of the reference category after accounting for other included variables. Driver age, vehicle age, and geographic area also provide useful information for risk segmentation.

After detecting substantial overdispersion in the initial Poisson model, Pearson scaling was applied to provide more appropriate estimates of statistical uncertainty. Out-of-sample validation showed reasonable model performance, with approximately 7,195 predicted claims compared with 7,241 actual claims on the test set, an aggregate underprediction of approximately 0.63%.

The model also demonstrates meaningful risk differentiation: the highest predicted-risk decile experienced approximately four times the observed claim frequency of the lowest predicted-risk decile. Overall, the analysis provides an interpretable foundation for insurance risk segmentation while recognizing that claim severity and other components would be required for a complete pricing analysis.

## 1. BonusMalus Is the Strongest Risk Indicator

BonusMalus is the strongest modeled predictor of claim frequency in this analysis.

Relative to the reference category of BonusMalus 50, the Pearson-scaled Poisson GLM estimates claim-frequency relativities of approximately:

- 1.61 for BonusMalus 51–75
- 2.25 for BonusMalus 76–100
- 5.17 for BonusMalus 101+

This means that, after accounting for the other variables included in the model, policy records in higher BonusMalus categories are associated with substantially higher expected claim frequency.

The pattern is also consistent with the earlier SQL segmentation and exploratory analysis, where observed claim frequency increased steadily as BonusMalus increased.

From a business perspective, BonusMalus appears to provide strong information for risk segmentation and could be an important variable when reviewing rating structure or underwriting segmentation. However, these results relate only to claim frequency and do not by themselves determine an indicated premium, since claim severity and other pricing components are not modeled here.

## 2. Driver Age Provides Meaningful Risk Segmentation

Driver age shows an important relationship with claim frequency. In the exploratory analysis, drivers aged 18–24 had the highest observed claim frequency at approximately 0.189, compared with roughly 0.09–0.11 for most other age groups.

After controlling for the other variables included in the model, drivers aged 25–34 had an estimated claim-frequency relativity of approximately 0.68 and drivers aged 35–44 had a relativity of approximately 0.86 relative to drivers aged 18–24. These estimates correspond to approximately 32% and 14% lower modeled claim frequency, respectively.

The relationship is not strictly monotonic across all age groups, and some older age-group coefficients are not statistically significant after adjusting for overdispersion. This demonstrates why multivariable modeling is important: observed differences between age groups can change after accounting for other risk characteristics.

From a business perspective, driver age may contribute useful information for risk segmentation, but its effect should be evaluated together with other rating characteristics rather than in isolation.

## 3. Vehicle Age Is Associated With Lower Claim Frequency

Vehicle age shows a consistent relationship with claim frequency. In the exploratory analysis, vehicles aged 0–10 years had an observed claim frequency of approximately 0.110, compared with 0.081 for vehicles aged 11–20 years and 0.054 for vehicles aged 21+ years.

The multivariable model shows a similar pattern. Relative to vehicles aged 0–10 years, vehicles aged 11–20 years have an estimated claim-frequency relativity of approximately 0.76, while vehicles aged 21+ years have a relativity of approximately 0.55. This corresponds to approximately 24% and 45% lower modeled claim frequency, respectively, after accounting for the other variables included in the model.

The consistency between the descriptive and multivariable results suggests that vehicle age provides useful information for risk segmentation. However, the analysis identifies an association rather than a causal relationship; it does not establish that increasing vehicle age itself reduces claim frequency.

From a business perspective, vehicle age may be useful when evaluating rating or underwriting segmentation, although its relationship with claim severity would also need to be considered before drawing conclusions about overall insurance cost.

## 4. Geographic Area Contributes to Risk Segmentation

Geographic area shows a clear relationship with claim frequency. In the exploratory analysis, observed claim frequency increased consistently from Area A to Area F, from approximately 0.082 in Area A to 0.139 in Area F.

The multivariable model indicates that geographic differences remain after accounting for the other included risk characteristics. Relative to Area A, the estimated claim-frequency relativities are approximately 1.09 for Area C, 1.18 for Area D, 1.24 for Area E, and 1.25 for Area F. Area B has a relativity of approximately 1.05, but its coefficient is not statistically significant at the 5% level after adjusting for overdispersion.

These results suggest that geographic area provides useful information for claim-frequency segmentation beyond the other variables included in the model.

Area is also strongly related to population density in this dataset. Therefore, the geographic results should be interpreted cautiously and should not be viewed as isolated causal effects.

From a business perspective, geographic segmentation may be useful when evaluating rating structures and portfolio risk, but additional analysis would be needed to determine how geography should be incorporated into a production pricing model.

## 5. Accounting for Overdispersion Changes Model Uncertainty

The initial Poisson GLM produced a Pearson dispersion statistic of approximately 2.70, substantially above the value of 1 expected under the standard Poisson variance assumption. This indicates that the claim counts exhibit more variability than the standard Poisson model assumes.

To account for this additional variability, the model was refitted using Pearson scaling. The scaling adjustment did not materially change the estimated coefficients or claim-frequency relativities, but it increased the estimated standard errors and widened the confidence intervals.

As a result, several coefficient estimates that appeared statistically significant under the standard Poisson model were no longer significant at the 5% level after adjusting for overdispersion.

From an actuarial perspective, this demonstrates the importance of evaluating model assumptions before relying on statistical significance. Ignoring overdispersion could lead to overstating the strength of evidence for some risk characteristics.

## 6. The Model Shows Reasonable Out-of-Sample Performance

The model was evaluated using a held-out test set containing 20% of the policy records. The model predicted approximately 7,195 claims compared with 7,241 actual claims, corresponding to an aggregate underprediction of approximately 0.63%.

Performance was also evaluated across predicted-risk deciles. Actual claim frequency generally increased as predicted risk increased, from approximately 0.052 in the lowest predicted-risk decile to 0.204 in the highest predicted-risk decile. Therefore, policy records in the highest predicted-risk decile experienced approximately four times the observed claim frequency of those in the lowest predicted-risk decile.

This pattern indicates that the model provides meaningful risk differentiation: policy records assigned higher predicted claim frequencies generally experienced higher actual claim frequencies.

Although predicted and actual frequencies are reasonably close across most risk groups, differences remain within individual deciles, particularly among some higher-risk segments. Therefore, the model demonstrates useful out-of-sample calibration and risk segmentation but should not be considered a production-ready pricing model.

## 7. Business and Pricing Implications

The analysis identifies several characteristics that may provide useful information for insurance risk segmentation. BonusMalus shows the strongest modeled relationship with claim frequency, while driver age, vehicle age, and geographic area also contribute information about differences in expected claim frequency.

From a pricing perspective, these results could help identify segments that may warrant further review. For example, higher BonusMalus categories are associated with substantially higher modeled claim frequency, while older vehicle categories are associated with lower modeled frequency. Geographic differences also remain after accounting for several other risk characteristics.

These findings could support further investigation of rating structures, underwriting segmentation, portfolio monitoring, and identification of higher- and lower-frequency risk groups.

However, claim frequency represents only one component of insurance loss cost. A complete pricing analysis would also need to consider claim severity, expenses, trends, credibility, regulatory requirements, and other business considerations. Therefore, the relativities estimated in this project should be interpreted as claim-frequency relationships rather than direct premium indications.

## 8. Limitations and Future Improvements

This project provides an interpretable analysis of auto insurance claim frequency, but several limitations should be considered.

First, the analysis models claim frequency only. Claim severity was not modeled, so the results cannot be used independently to determine total expected loss cost or indicated premiums.

Second, the Poisson model exhibited overdispersion. Pearson scaling was used to adjust the estimated uncertainty, but it does not change the underlying Poisson mean structure or eliminate the additional variability in the claim data.

Third, the model uses a limited set of predictors and does not include interaction terms. Additional relationships between risk characteristics may therefore remain unexplained. Geographic variables such as Area and Density also show substantial overlap, which limits the ability to interpret their effects independently.

Finally, the model was developed as an interpretable portfolio analysis rather than a production pricing model. Future work could evaluate alternative model specifications, additional interactions, claim severity modeling, and more extensive model validation.

Despite these limitations, the analysis demonstrates meaningful risk segmentation and reasonable out-of-sample performance, providing a useful foundation for further actuarial modeling.
